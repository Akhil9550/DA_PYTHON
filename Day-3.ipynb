```python
from decimal import Decimal
monthly_salary=Decimal("1000.00")
annual_salary=monthly_salary*12
print("Annual Salary:", annual_salary)
print("monthly_salary:", monthly_salary)
print("type of monthly_salary:", type(monthly_salary))
```

```python
from decimal import Decimal
integer_value = Decimal(10)
float_value = Decimal("10.5")
complex_number = 10 + 5j
print("integer_value:", integer_value)
print("Datatype:", type(integer_value))
print("float_value:", float_value)
print("Datatype:", type(float_value))
print("complex_number:", complex_number)
print("Datatype:", type(complex_number))
```
